```python
import copy
import numpy as np
import matplotlib.pyplot as plt
from nonlinear_approximator import (
    activations,
    inference,
    model,
    params,
    training, 
)
%matplotlib inline
```

```python

config = params.RegressionParams(
    width=50,
    depth=100,
    input_dimension=2,
    transform_type=activations.TransformType.TENT,
    transform_params= params.TentParams(mu=2.0),
    output_dimension=1,
)
plt.title("Tent Map Nonlinear Activation")
xs=np.linspace(0,1,1000)
for j in range(4): 
    acts = activations.neuron_activation_curve(xs, j+1, config )
    plt.plot(xs, acts, label=f"{j+1}")
plt.legend()
plt.xlabel("x")
plt.ylabel("${T^k}$(x)")
plt.ion()
plt.show()
```
